## Imports

```python
import pandas as pd
```

## Creating a DataFrame

A DataFrame is a table. It contains an array of individual entries, each of which has a certain value. Each entry corresponds to a row (or record) and a column.

The most common way to create a DataFrame is from a Python dictionary.

```python
# Create a DataFrame from a dictionary
# Keys become column names, values become column data
df = pd.DataFrame({'Yes': [50, 21], 'No': [131, 2]})
df
```

```python
# DataFrame entries can be strings too
df_strings = pd.DataFrame({
    'Bob': ['I liked it.', 'It was awful.'],
    'Sue': ['Pretty good.', 'Bland.']
})
df_strings
```

```python
# Use the index parameter to set custom row labels
df_indexed = pd.DataFrame(
    {'Bob': ['I liked it.', 'It was awful.'], 'Sue': ['Pretty good.', 'Bland.']},
    index=['Product A', 'Product B']
)
df_indexed
```

## Creating a Series

A Series is a sequence of data values — essentially a single column of a DataFrame.
You can create a Series from a list.

```python
# Create a Series from a list
s = pd.Series([1, 2, 3, 4, 5])
s
```

```python
# A Series is essentially a single column of a DataFrame
# You can assign a name and custom index
s_named = pd.Series(
    [30, 35, 40],
    index=['2015 Sales', '2016 Sales', '2017 Sales'],
    name='Product A'
)
s_named
```

## Reading Data Files

Being able to create a DataFrame is handy. But, most of the time, we won't actually be creating our own data by hand. Instead, we'll be working with data that already exists.

Data can be stored in any of a number of different forms and formats. By far the most basic of these is the humble CSV file.

```python
# First, let's create a sample CSV file to work with
import io

csv_data = """country,variety,points,price
Italy,White Blend,87,16
Portugal,Portuguese White,87,15
US,Pinot Gris,87,14
US,Riesling,87,11
Spain,Nerello Mascalese,87,11
Italy,Nerello Mascalese,87,40
France,Rosé,88,15
Germany,Riesling,88,12
Argentina,Malbec,88,20
US,Cabernet Sauvignon,88,65
"""

with open('/tmp/wine_reviews.csv', 'w') as f:
    f.write(csv_data)

print("Sample CSV file created at /tmp/wine_reviews.csv")
```

```python
# Read a CSV file using pd.read_csv()
wine_reviews = pd.read_csv('/tmp/wine_reviews.csv')
wine_reviews
```

```python
# Use index_col to set one of the columns as the row index
wine_reviews_indexed = pd.read_csv('/tmp/wine_reviews.csv', index_col=0)
wine_reviews_indexed
```

## Inspecting a DataFrame

Once you have a DataFrame, you'll want to get a quick sense of what it looks like.

```python
# shape returns a tuple of (rows, columns)
wine_reviews.shape
```

```python
# head() returns the first 5 rows by default
wine_reviews.head()
```

```python
# dtypes shows the data type of each column
wine_reviews.dtypes
```

```python
# info() provides a concise summary of the DataFrame
wine_reviews.info()
```

## Accessing Data

There are two primary methods for accessing data in a DataFrame:
- **`loc`** – label-based selection
- **`iloc`** – integer position-based selection

```python
# Access a column by name
wine_reviews['country']
```

```python
# iloc selects by integer position (like Python list indexing)
# Select the first row
wine_reviews.iloc[0]
```

```python
# iloc with [row, column] indexing
# Select the value in row 0, column 0
wine_reviews.iloc[0, 0]
```

```python
# Select the first column using iloc
wine_reviews.iloc[:, 0]
```

```python
# loc selects by label
# Select row 0 using loc
wine_reviews.loc[0, 'country']
```

```python
# loc is useful for selecting specific columns by name across all rows
wine_reviews.loc[:, ['country', 'variety']]
```

## Writing Data Files

You can save a DataFrame to a CSV file using the `to_csv()` method.
Set `index=False` to avoid writing the row index as an extra column.

```python
# Write a DataFrame to a CSV file
wine_reviews.to_csv('/tmp/wine_reviews_out.csv', index=False)
print("DataFrame saved to /tmp/wine_reviews_out.csv")
```

```python
# Verify the file was written correctly by reading it back
verify_df = pd.read_csv('/tmp/wine_reviews_out.csv')
verify_df.head()
```

## Summary

| Task | Method |
|------|--------|
| Create DataFrame from dict | `pd.DataFrame({'col': [values]})` |
| Create a Series | `pd.Series([values], name='name')` |
| Read CSV | `pd.read_csv('file.csv')` |
| Read CSV with index column | `pd.read_csv('file.csv', index_col=0)` |
| Inspect shape | `df.shape` |
| Preview first rows | `df.head()` |
| Check data types | `df.dtypes` |
